```python
dfn=df[['traveltime','studytime']]
x=dfn['traveltime']
y=dfn['studytime']
sbs.lineplot(x=x,y=y,dashes=True)
plt.show()
```

```python
# correlation analysis
correlation=dfn.corr()
print(correlation)
sbs.heatmap(correlation,cmap="YlGnBu")
plt.show()

```

```python
covar=dfn.cov()
print(covar)
sbs.heatmap(covar,cmap="YlGnBu")
plt.show()
```
